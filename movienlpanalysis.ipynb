## Text Cleaning (NLP):

```python
import nltk
import re

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
print(dfmovies['StoryLine'].head())  # Check the first few rows
print(dfmovies['StoryLine'].isnull().sum())  # Check for null values
print(dfmovies['StoryLine'].apply(type).value_counts())  # Check data types
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def NLP_preprocess(text):
    if not isinstance(text, str) or text.strip() == "":  # Handle non-string or empty values
        return ""
    try:
        text = text.lower()
        text = re.sub(r'[^a-z\s]', '', text)  # Remove non-alphabetic characters
        tokens = word_tokenize(text)  # Tokenize the text
        tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]  # Lemmatize and remove stopwords
        return ' '.join(tokens)  # Join tokens back into a single string
    except Exception as e:
        print(f"Error processing text: {text}")
        print(f"Error: {e}")
        return ""
```

```python
dfmovies['Cleaned_Storyline'] = dfmovies['StoryLine'].apply(NLP_preprocess)
```

## Text Representation:

```python
from sklearn.feature_extraction.text import CountVectorizer
count_vectorizer = CountVectorizer(max_features=1000)
```

```python
count_matrix = count_vectorizer.fit_transform(dfmovies['Cleaned_Storyline'])
```

```python
dfmoviescount = pd.DataFrame(count_matrix.toarray(), columns=count_vectorizer.get_feature_names_out())
```

```python
dfmovies['Count_Vector'] = list(dfmoviescount.values)
```

## Cosine Similarity:

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
count_matrix = np.array(dfmovies['Count_Vector'].to_list())
count_matrix
```

```python
cosine_sim_matrix = cosine_similarity(count_matrix)
cosine_sim_matrix
```

```python
def get_ranked_movies(index, cosine_sim_matrix, movies_df, top_n=5):
    sim_scores = list(enumerate(cosine_sim_matrix[index]))
    sorted_sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    top_movies = sorted_sim_scores[1:top_n+1]
    ranked_movies = [dfmovies.iloc[i]['MovieName'] for i, score in top_movies]
    return ranked_movies
```

```python
dfmovies['Ranked_Movies'] = [
    get_ranked_movies(i, cosine_sim_matrix, dfmovies) for i in range(len(dfmovies))
]
dfmovies[['MovieName', 'Ranked_Movies']].head(10)
```
